# Visual Story

This demo creates images, color maps, and SVG output without downloading any assets. It is a tour of the media paths that make notebooks feel like illustrated reports.


```python
import numpy as np
from IPython.display import display
from PIL import Image, ImageDraw

width, height = 720, 300
x = np.linspace(0, 1, width)
y = np.linspace(0, 1, height)[:, None]
red = (40 + 150 * x).astype("uint8")
green = (90 + 120 * (1 - y)).astype("uint8")
blue = (120 + 80 * np.sin(2 * np.pi * (x + y)) ** 2).astype("uint8")
canvas = np.dstack([
    np.tile(red, (height, 1)),
    np.tile(green, (1, width)),
    blue,
])
image = Image.fromarray(canvas, "RGB")
draw = ImageDraw.Draw(image)
draw.rectangle((38, 42, 334, 124), fill=(255, 255, 255), outline=(8, 127, 91), width=3)
draw.text((58, 64), "Generated notebook image", fill=(23, 33, 28))
draw.text((58, 92), "served lazily by the viewer", fill=(23, 33, 28))
display(image)
```

## Matplotlib color fields

Generated image outputs are stored as local assets. The browser receives an asset URL instead of a base64 blob in the main section response.


```python
import matplotlib.pyplot as plt

xx = np.linspace(-3, 3, 240)
yy = np.linspace(-2, 2, 180)
X, Y = np.meshgrid(xx, yy)
field_a = np.sin(X ** 2 + Y ** 2)
field_b = np.cos(2 * X) * np.sin(3 * Y)

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
for ax, field, title, cmap in zip(axes, [field_a, field_b], ["radial waves", "interference"], ["viridis", "magma"]):
    im = ax.imshow(field, cmap=cmap, origin="lower", extent=[xx.min(), xx.max(), yy.min(), yy.max()])
    ax.set_title(title)
    ax.set_xlabel("x")
    ax.set_ylabel("y")
    fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
fig.tight_layout()
plt.show()
```
