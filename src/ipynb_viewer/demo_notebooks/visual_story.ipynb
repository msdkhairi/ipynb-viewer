# Visual Story

This demo creates image outputs from arrays and small vector graphics without loading external assets.

```python
import numpy as np
from PIL import Image, ImageDraw
from IPython.display import display

rng = np.random.default_rng(9)
tiles = []
for index in range(9):
    base = np.linspace(0, 1, 96)
    xx, yy = np.meshgrid(base, base)
    wave = np.sin((index + 2) * xx * np.pi) * np.cos((index + 1) * yy * np.pi)
    noise = 0.18 * rng.normal(size=wave.shape)
    red = np.clip(120 + 95 * wave + 20 * index, 0, 255)
    green = np.clip(150 + 90 * yy + 60 * noise, 0, 255)
    blue = np.clip(170 + 80 * xx - 70 * wave, 0, 255)
    tile = np.dstack([red, green, blue]).astype('uint8')
    tiles.append(Image.fromarray(tile))

canvas = Image.new('RGB', (3 * 104, 3 * 104), 'white')
for index, tile in enumerate(tiles):
    x = (index % 3) * 104
    y = (index // 3) * 104
    canvas.paste(tile, (x, y))
display(canvas)
```

## Matplotlib Color Fields

Generated arrays become responsive image outputs in the viewer.

```python
import matplotlib.pyplot as plt

grid = np.linspace(-2.8, 2.8, 260)
xx, yy = np.meshgrid(grid, grid)
field = np.sin(xx ** 2 + yy ** 2) / (1 + 0.16 * (xx ** 2 + yy ** 2))

fig, ax = plt.subplots(figsize=(5.4, 4.5))
im = ax.imshow(field, cmap='magma', origin='lower', extent=[grid.min(), grid.max(), grid.min(), grid.max()])
ax.contour(xx, yy, field, levels=10, colors='white', linewidths=0.35, alpha=0.7)
ax.set_title('Generated scalar field')
ax.set_xticks([])
ax.set_yticks([])
fig.colorbar(im, ax=ax, shrink=0.76)
fig.tight_layout()
plt.show()
```
