## A Small Plot

The viewer keeps the chart output visible while the code stays tucked away.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(7)
x = np.linspace(0, 2 * np.pi, 240)
y = np.sin(x) + 0.12 * rng.normal(size=x.size)

fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y, color='#087f5b', linewidth=2)
ax.plot(x, np.sin(x), color='#1d4ed8', linewidth=1.5, alpha=0.65)
ax.set_title('A noisy sine wave')
ax.set_xlabel('phase')
ax.set_ylabel('amplitude')
ax.grid(alpha=0.22)
fig.tight_layout()
plt.show()
```
