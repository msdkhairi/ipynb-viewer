# Quickstart Tour

This notebook is the first-run tour for `ipynb-local-viewer`. It keeps the story small on purpose: headings become the outline, outputs appear before code, code stays collapsed until you ask for it, and rich media is streamed lazily by the local viewer.


```python
from datetime import datetime

import numpy as np
import pandas as pd
from IPython.display import HTML, display

rng = np.random.default_rng(7)
sections = pd.DataFrame(
    {
        "viewer capability": [
            "Markdown outline",
            "Output-first reading",
            "Collapsed source",
            "Lazy media assets",
            "Sanitized HTML",
        ],
        "why it matters": [
            "Jump between notebook sections like a report.",
            "Read results before diving into implementation details.",
            "Keep long analysis cells available but quiet.",
            "Large images and videos do not bloat the first page payload.",
            "Notebook HTML is cleaned before it reaches the browser.",
        ],
        "status": ["ready", "ready", "ready", "ready", "ready"],
    }
)

print(f"Live demo started at {datetime.utcnow():%Y-%m-%d %H:%M:%S} UTC")
display(sections)

display(HTML("""
<table style="border-collapse:collapse;width:100%;max-width:760px">
  <tr>
    <th style="text-align:left;padding:8px;border:1px solid #dbe3dc">Reading mode</th>
    <th style="text-align:left;padding:8px;border:1px solid #dbe3dc">Default behavior</th>
  </tr>
  <tr>
    <td style="padding:8px;border:1px solid #dbe3dc;font-weight:700">Saved notebooks</td>
    <td style="padding:8px;border:1px solid #dbe3dc">Read only; code is not executed.</td>
  </tr>
  <tr>
    <td style="padding:8px;border:1px solid #dbe3dc;font-weight:700">Packaged demos</td>
    <td style="padding:8px;border:1px solid #dbe3dc">Executed only when launched with <code>--demo</code>.</td>
  </tr>
</table>
"""))
```

## Output-first analysis

The viewer puts results before the source code for each cell. This makes notebooks feel more like polished reports while preserving the exact code behind every result.


```python
import matplotlib.pyplot as plt

hours = np.arange(0, 24)
read_time = 18 + 8 * np.sin((hours - 6) / 24 * 2 * np.pi) + rng.normal(0, 1.2, len(hours))
media_cost = 4 + 12 * np.exp(-((hours - 13) ** 2) / 32)

fig, ax = plt.subplots(figsize=(9, 4.2))
ax.plot(hours, read_time, color="#087f5b", linewidth=2.5, label="reader activity")
ax.fill_between(hours, media_cost, color="#1d4ed8", alpha=0.18, label="lazy media demand")
ax.set_title("A compact notebook result rendered as a report section")
ax.set_xlabel("hour of day")
ax.set_ylabel("relative activity")
ax.grid(True, alpha=0.25)
ax.legend(loc="upper right")
fig.tight_layout()
plt.show()
```

## Tables, HTML, and lazy media callouts

Small HTML blocks are sanitized and rendered inline. Heavier image or video outputs are stored as local assets and loaded through `/api/asset/...` only when the reader needs them.


```python
summary = pd.DataFrame(
    {
        "output type": ["stdout", "pandas table", "matplotlib image", "sanitized HTML", "source code"],
        "viewer treatment": [
            "terminal-style block",
            "scrollable rich HTML table",
            "lazy image asset",
            "cleaned before rendering",
            "collapsed, syntax highlighted details",
        ],
    }
)
display(summary)

html = """
<div style="border:1px solid #dbe3dc;padding:14px;background-color:#fbfcfb;max-width:760px">
  <strong>Local-first default:</strong>
  this reader opens saved notebooks without executing their code. Demo execution is explicit,
  isolated to packaged examples, and visible through the live status pill.
</div>
"""
display(HTML(html))
```
