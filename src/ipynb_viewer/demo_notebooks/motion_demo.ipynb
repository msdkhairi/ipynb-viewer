# Motion Demo

This notebook generates a tiny animation at runtime. The output is stored in the executed notebook and served lazily by the viewer.

```python
import io
import numpy as np
from PIL import Image, ImageDraw
from IPython.display import Image as IPythonImage, display

frames = []
width, height = 320, 180
for frame_index in range(28):
    frame = Image.new('RGB', (width, height), '#111827')
    draw = ImageDraw.Draw(frame)
    for offset in range(0, width, 28):
        x = (offset + frame_index * 5) % (width + 28) - 28
        draw.line((x, 0, x + 85, height), fill='#1f2937', width=3)
    phase = frame_index / 27
    cx = int(44 + phase * (width - 88))
    cy = int(height / 2 + 34 * np.sin(phase * 2 * np.pi))
    draw.ellipse((cx - 18, cy - 18, cx + 18, cy + 18), fill='#34d399')
    draw.ellipse((cx - 7, cy - 7, cx + 7, cy + 7), fill='#ecfdf5')
    draw.text((14, 14), 'generated GIF output', fill='#dbeafe')
    frames.append(frame)

buffer = io.BytesIO()
frames[0].save(buffer, format='GIF', save_all=True, append_images=frames[1:], duration=65, loop=0)
display(IPythonImage(data=buffer.getvalue(), format='gif'))
```

## Tiny MP4 Video

A small packaged MP4 is embedded by IPython, then the viewer extracts it into a streamed asset URL.

```python
from importlib import resources
from IPython.display import Video, display

video_resource = resources.files('ipynb_viewer.demo_notebooks').joinpath('assets/tiny_motion.mp4')
with resources.as_file(video_resource) as video_path:
    display(Video(filename=str(video_path), embed=True, mimetype='video/mp4', width=320))
```
