# Motion Demo

A lightweight media notebook for checking video and animated output paths. It uses a tiny packaged MP4 so the demo stays offline and quick to run.


```python
import numpy as np
import pandas as pd
from IPython.display import display
from PIL import Image, ImageDraw

frames = []
for step in range(6):
    image = Image.new("RGB", (150, 92), (247, 248, 246))
    draw = ImageDraw.Draw(image)
    x = 18 + step * 20
    draw.rectangle((10, 10, 140, 82), outline=(219, 227, 220), width=2)
    draw.ellipse((x, 30, x + 34, 64), fill=(8, 127, 91))
    draw.text((16, 66), f"frame {step + 1}", fill=(23, 33, 28))
    frames.append(image)

strip = Image.new("RGB", (150 * len(frames), 92), "white")
for index, frame in enumerate(frames):
    strip.paste(frame, (index * 150, 0))
display(strip)

catalog = pd.DataFrame(
    {
        "asset": ["generated frame strip", "packaged MP4", "summary table"],
        "viewer path": ["image asset", "video asset", "HTML/table output"],
    }
)
display(catalog)
```

## Tiny MP4 video

The viewer extracts embedded video data into a local cache file and streams it through the asset API, keeping the section payload small.


```python
from importlib import resources
from IPython.display import Video, display

video_resource = resources.files("ipynb_viewer.demo_notebooks").joinpath("assets/tiny_motion.mp4")
with resources.as_file(video_resource) as video_path:
    display(Video(filename=str(video_path), embed=True, mimetype="video/mp4", width=520))
```
